## 1. Терминал

Терминал - это способ давать компьютеру команды текстом. Он нужен не потому, что «так делают настоящие программисты», а потому что целый класс инструментов - установка библиотек, виртуальные окружения, git, docker, запуск серверов - управляется текстом и не имеет кнопки.

Минимум, который надо освоить:

| Команда | Что делает |
|---|---|
| `pwd` | где я сейчас нахожусь |
| `ls` | что лежит в текущей папке (`ls -a` - включая скрытые файлы) |
| `cd путь` | перейти в папку (`cd ..` - на уровень вверх) |
| `cat файл` | показать содержимое файла |
| `mkdir имя` | создать папку |
| `cp` / `mv` / `rm` | копировать / переименовать или переместить / удалить |
| `Tab` | автодополнение имени - жмите постоянно, это защита от опечаток |
| `↑` | предыдущая команда |

Две вещи, на которых спотыкаются почти все.

**Пути бывают абсолютные и относительные.** `/Users/ivan/projects/ml_learning_course/data/file.csv` - абсолютный, работает откуда угодно. `data/file.csv` - относительный, он отсчитывается от папки, в которой вы находитесь **прямо сейчас**. Классическая ошибка: в коде написано `data/file.csv`, а ноутбук запущен из другой папки - файл не находится. Отсюда привычка: запускать всё из корня проекта.

**Скрытые файлы начинаются с точки.** `.env`, `.gitignore`, `.venv` - обычный `ls` их не покажет, нужен `ls -a`. Файловый менеджер их тоже прячет. Именно поэтому люди «теряют» `.env` и удивляются, что настройки не подхватились.

## 2. Установка и окружение

Подробная инструкция по операционным системам - [docs/setup.md](../docs/setup.md). Здесь - смысл каждого шага, чтобы вы понимали, что делаете.

**Python 3.11.** Версию фиксируем: код с 3.11 может не запуститься на 3.9, а `numpy 2.1` на 3.9 просто не установится. В корне лежит `.python-version` - напоминание, какая версия нужна проекту.

**Виртуальное окружение.** По умолчанию `pip` ставит библиотеки глобально, в системный Python. Пока проект один - незаметно. Как только их два, начинается конфликт: проекту A нужен `pandas 1.5`, проекту B - `pandas 2.2`. Установить обе версии глобально нельзя; обновили ради B - сломали A. Виртуальное окружение - это отдельная папка `.venv` со своей копией Python и своим набором библиотек:

```bash
python3 -m venv .venv          # создать
source .venv/bin/activate      # включить (macOS / Linux / WSL)
```

После активации в строке терминала появляется `(.venv)`. **Это единственный признак, по которому видно, что вы внутри.** Нет его - `pip install` поставит библиотеку мимо проекта, в систему. Ровно так возникает «я же установил, а он пишет, что модуля нет».

**Зависимости.** `pip install -r requirements.txt` ставит всё нужное одной командой. Сама папка `.venv` в git не уезжает - она весит сотни мегабайт и у каждой ОС своя. В git уезжает только список:

```
pandas==2.2.3
numpy==2.1.3
```

Версии зафиксированы жёстко, через `==`. Если написать просто `pandas`, сегодня встанет 2.2.3, а через полгода - 3.0, где часть методов переименована, и проект перестанет запускаться сам по себе, без единой правки в коде. Правило: **поставили библиотеку - сразу дописали её в `requirements.txt` с версией.**

**Ядро Jupyter.** Отдельная ловушка. Jupyter - самостоятельная программа, и она сама решает, каким Python выполнять ваш код. Этот Python называется **ядром** (kernel), и по умолчанию он не тот, что в вашем `.venv`. Поэтому окружение регистрируют как ядро:

```bash
python -m ipykernel install --user --name ml-course --display-name "Python (ml-course)"
```

и в ноутбуке выбирают ядро **Python (ml-course)** (справа сверху). Большая часть ошибок `ModuleNotFoundError` на этом модуле - это именно неправильно выбранное ядро, а вовсе не отсутствующая библиотека. Проверка занимает одну строку и должна стоять первой ячейкой любого вашего ноутбука:

```python
import sys; print(sys.executable)
```

Путь обязан вести внутрь `.venv` вашего проекта. Ведёт в `/usr/bin/python3` или в `anaconda3` - вы не в своём окружении.

**Редактор.** VS Code или Cursor (это VS Code со встроенным AI). Ставим расширения **Python**, **Jupyter**, на Windows дополнительно **WSL**. После установки выбираем интерпретатор: `Cmd/Ctrl+Shift+P → Python: Select Interpreter →` тот, что внутри `.venv` вашего проекта. Ноутбуки открываются прямо в редакторе, отдельный Jupyter в браузере не обязателен.

## 3. Git

Git решает три задачи: хранит историю изменений, позволяет откатиться и даёт работать вместе. Для вас сейчас важнее первая и третья: история делает проект предъявляемым, а через pull request вас будут ревьюить.

Минимальный цикл, который вы повторяете каждую домашку:

```bash
git status                       # что изменилось (смотреть ПЕРЕД каждым коммитом)
git checkout -b hw01             # создать ветку под задачу и перейти в неё
git add .                        # отобрать изменения к фиксации
git commit -m "hw01: окружение"  # зафиксировать снимок с сообщением
git push -u origin hw01          # отправить ветку на GitHub
```

Ещё несколько команд, которые понадобятся:

| Команда | Что делает |
|---|---|
| `git clone <url>` | скачать репозиторий себе |
| `git log --oneline` | список коммитов |
| `git diff` | что именно изменилось в файлах |
| `git checkout main` | вернуться в основную ветку |
| `git pull upstream main` | забрать новые задания от преподавателя |
| `git remote -v` | посмотреть, с какими репозиториями вы связаны |

**Ветка** - отдельная линия работы; каждая домашка делается в своей, `main` остаётся чистым. **Коммит** - зафиксированный снимок с сообщением; сообщение пишется про то, *что сделано*, а не «фикс», «апдейт», «ещё раз». **Pull request** - предложение влить ветку и одновременно место, где проверяющий пишет комментарии к конкретным строкам.

> 🎮 Чтобы ветки и коммиты стали наглядными, пройдите тренажёр **[learngitbranching.js.org](https://learngitbranching.js.org)** (есть русский язык). Раздел «Основы», уровни 1-4 - это минут двадцать, и они снимают большую часть страха перед git.

**`.gitignore`** - список того, что git обязан игнорировать: секреты (`.env`), мусор окружения (`.venv/`, `__pycache__/`, `.ipynb_checkpoints/`), данные (`data/`). Проверить, что файл реально игнорируется: `git check-ignore -v .env`.

**Ноутбук в git.** Технически `.ipynb` - это JSON, внутри которого лежат и код, и все выводы ячеек, включая картинки в base64. Изменение одной строки даёт дифф на тысячу строк, и ревью становится невозможным. Лечится `nbstripout`: ставится один раз командой `nbstripout --install` и дальше сам вырезает выводы при коммите.

**Секреты.** Пароль в коде - главная ошибка, которую тут можно совершить. Git запоминает всё навсегда: закоммитили пароль, стёрли следующим коммитом - он остаётся в истории и достаётся одной командой. Публичные репозитории сканируются ботами за минуты. Поэтому доступы живут в `.env`, который перечислен в `.gitignore`, а рядом лежит `.env.example` с пустыми значениями - по нему видно, какие переменные нужны проекту. Привычка на всю жизнь: **перед каждым коммитом смотреть `git status` и убеждаться, что `.env` там нет.**

В этой домашке вы работаете с **публичным ClickHouse**, где лежат открытые датасеты. Регистрация, VPN и создание таблиц не нужны - доступы уже прописаны в `.env.example`, достаточно скопировать файл.

| Колонка | Что внутри |
|---|---|
| `trip_id` | идентификатор поездки |
| `pickup_datetime`, `dropoff_datetime` | время начала и конца поездки |
| `pickup_longitude`, `pickup_latitude` | координаты посадки |

| `payment_type` | способ оплаты: `CSH` наличные, `CRE` карта, `NOC`/`DIS` прочее |

Подключение уже написано за вас в [`src/db.py`](../src/db.py) - там сорок строк, и вы должны понимать, что в них происходит. Файл читает `.env`, собирает клиент ClickHouse и отдаёт его наружу. В ноутбуке это выглядит так:

```python
from src.db import get_client, query_df

client = get_client()
print(client.query("SELECT version()").result_rows)

df = query_df("SELECT * FROM nyc_taxi.trips_small LIMIT 10")
```

`query_df` - тот же запрос, но результат сразу приезжает в `pandas.DataFrame`. Никаких хостов и паролей в ячейках нет и быть не должно: они приходят из `.env`.

Данные сырые, как в реальной жизни: есть поездки с нулевым расстоянием, отрицательными суммами и пустым районом. Это не баг датасета, это нормальное состояние данных, и фильтровать их - часть вашей работы.

> ⚠️ База публичная и на ней сидит весь интернет. Ставьте `LIMIT`, не тяните миллионы строк в `DataFrame` и агрегируйте на стороне ClickHouse, а не в pandas. Это, кстати, и есть правильный рефлекс: считает база, python получает готовый результат.

Если хотите работать со своим сервисом ClickHouse Cloud с SQL-модуля - впишите в `.env` его доступы вместо публичных. Но датасета такси там не будет, его придётся заливать самому.
